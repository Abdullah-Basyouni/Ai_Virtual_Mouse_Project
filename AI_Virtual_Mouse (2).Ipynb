### Install Library

```python
import cv2
import mediapipe as mp
import autopy as ap
import numpy as np 
```

### Ai Vitual Mouse

```python
camera = cv2.VideoCapture(0)
mp_hands = mp.solutions.hands
mp_draw = mp.solutions.drawing_utils
mainHand = mp_hands.Hands()

#Dimentions
WScr , HScr = ap.screen.size()
px , py = 0 , 0
Cx , Cy = 0 , 0

def extractLandmarks(colorImg  , frame):
    Landmarks = []
    
    landmarkspoints = mainHand.process(colorImg)
    landmarksCheck = landmarkspoints.multi_hand_landmarks
    
    if landmarksCheck:
        for hand in landmarksCheck:
            for index , landmark in enumerate(hand.landmark):
                mp_draw.draw_landmarks(frame , hand , mp_hands.HAND_CONNECTIONS)
                #Postions
                h , w , c = frame.shape
                centerX , centerY = int(landmark.x * w) , int(landmark.y * h)
                
                Landmarks.append([index , centerX  , centerY])
                
    return Landmarks

def fingerstip(landlist):
    fingers =  []
    tipId  = [4 , 8 , 12 , 16 , 20]
    
    #thumb up 
    if landlist[tipId[0]][1] > landlist[tipId[0] - 1 ][1]:
        fingers.append(0)
    else :
        fingers.append(1)
        
    
    # Other Fingers
    for i in range(1,5):
        if landlist[tipId[i]][2] < landlist[tipId[i] - 3 ][2]:
            fingers.append(1)
            
        else : 
            fingers.append(0)
            
    return fingers
                
                
                
                
                
                
while True:
    reg , frame = camera.read()
    
    frame =  cv2.flip(frame , 1)
    
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    
    Llist = extractLandmarks(rgb_frame , frame)
    
    if len(Llist) != 0 :
        x1 , y1 = Llist[8][1:]
        
        fingers = fingerstip(Llist)
        
        if fingers[1] == 1 and fingers[2] == 0:
            h , w , c = frame.shape
            
            X2 = np.interp(x1 , (0 , w)  , (0 , WScr))
            Y2 = np.interp(y1 , (0 , h) , (0 , HScr))
            
            Cx = px + (X2 - px) / 7 
            Cy = py + (Y2 - py) / 7
            
            ap.mouse.move(Cx , Cy)
            px , py = Cx , Cy
            
        if fingers[0] == 1 and fingers[2] == 0:
            ap.mouse.click()
            
    cv2.imshow("Camera", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break
    

camera.release()
cv2.destroyAllWindows()
             
```
